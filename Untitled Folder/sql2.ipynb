```python
import pandas as pd
import sqlite3

# 1. Fake Data banaya
sql_data = {
    'Emp_ID': [101, 102, 103, 104, 105, 106, 107],
    'Name': ['Ravi', 'Priya', 'Kiran', 'Amit', 'Neha', 'Rahul', 'Sonia'],
    'Department': ['IT', 'HR', 'IT', 'Sales', 'HR', 'IT', 'Sales'],
    'Salary': [55000, 62000, 48000, 75000, 50000, 90000, 45000],
    'Joining_Year': [2020, 2019, 2021, 2018, 2022, 2018, 2023]
}
emp_df = pd.DataFrame(sql_data)

# 2. SQL Engine (Connection) banaya memory mein
conn = sqlite3.connect(':memory:')

# 3. Pandas DataFrame ko SQL Table ('employees') mein convert kar diya
emp_df.to_sql('employees', conn, index=False)

print("SQL Database Ready! Table name is 'employees'.")
```

```python
query1="""
SELECT Name,Department
FROM employees
"""
pd.read_sql_query(query1,conn)
```

```python
query2="""
SELECT Name,Department,Salary
FROM employees
WHERE Salary>60000
ORDER BY Salary DESC
"""
pd.read_sql_query(query2,conn)
```

```python
query3="""
SELECT Name,Department,Salary
FROM employees
WHERE Department=='IT' AND Salary>50000
ORDER BY Salary DESC
"""
pd.read_sql_query(query3,conn)
```

```python
query4="""
SELECT *
FROM employees
ORDER BY Salary DESC
"""
pd.read_sql_query(query4,conn)
```

```python
query5="""
SELECT *
FROM employees
ORDER BY Joining_Year ASC
LIMIT 2
"""
pd.read_sql_query(query5,conn)
```

```python
query6="""
SELECT COUNT(Emp_ID) AS Total_Employees 
FROM employees
"""
pd.read_sql_query(query6,conn)
```

```python
query7="""
SELECT COUNT(Emp_ID) AS Total_Employees, Department
FROM employees
GROUP BY Department
"""
pd.read_sql_query(query7,conn)
```

```python
query8="""
SELECT Department, SUM(Salary) AS Total_sal
FROM employees
GROUP BY Department
HAVING SUM(Salary)
ORDER BY SUM(Salary) DESC
"""
pd.read_sql_query(query8,conn)
```

```python
query9="""
SELECT Department, MAX(Salary) AS Highest_Salary,Name
FROM employees
WHERE Department='IT'
ORDER BY MAX(Salary) DESC
"""
pd.read_sql_query(query9,conn)
```

```python
query10="""
SELECT Department,AVG(Salary) AS Average_Salary
FROM employees
GROUP BY Department
HAVING AVG(Salary)>55000
ORDER BY AVG(Salary) DESC
"""
pd.read_sql_query(query10,conn)
```

```python
import pandas as pd
import sqlite3

# 1. Customers Table (Total 4 customers)
data_cust = {
    'C_ID': [1, 2, 3, 4], 
    'Name': ['Amit', 'Neha', 'Rahul', 'Pooja'], 
    'City': ['Delhi', 'Mumbai', 'Pune', 'Delhi']
}
# 2. Orders Table (Total 5 orders)
data_ord = {
    'O_ID': [101, 102, 103, 104, 105], 
    'C_ID': [1, 2, 1, 5, 2], # Dhyan se dekh, C_ID 5 table mein nahi hai (Guest order)
    'Product': ['Laptop', 'Phone', 'Mouse', 'Tablet', 'Keyboard'], 
    'Amount': [50000, 20000, 1000, 15000, 1500]
}

df_cust = pd.DataFrame(data_cust)
df_ord = pd.DataFrame(data_ord)

# Database Setup
conn = sqlite3.connect(':memory:')
df_cust.to_sql('customers', conn, index=False)
df_ord.to_sql('orders', conn, index=False)
print("Tables 'customers' and 'orders' are ready!")
```

```python
query11="""
SELECT customers.Name,orders.Product,orders.C_ID
FROM customers
LEFT JOIN orders
ON customers.C_ID=orders.C_ID
"""
pd.read_sql_query(query11,conn)
```

```python
query12="""
SELECT customers.Name,customers.C_ID,orders.Product
FROM customers
INNER JOIN orders
ON customers.C_ID=orders.C_ID
"""
pd.read_sql_query(query12,conn)
```

```python
query13="""
SELECT customers.Name,customers.City,orders.Product,orders.C_ID
FROM orders
LEFT JOIN customers
ON customers.C_ID=orders.C_ID
WHERE City='Delhi'
"""
pd.read_sql_query(query13,conn)
```

```python
query14="""
SELECT customers.Name,customers.C_ID,SUM(orders.Amount) AS Total_amount,orders.Product
FROM orders
LEFT JOIN customers
ON customers.C_ID=orders.C_ID
GROUP BY customers.Name
ORDER BY SUM(orders.Amount) DESC
"""
pd.read_sql_query(query14,conn)
```
