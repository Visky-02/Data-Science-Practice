```python
WITH First_CTE AS (
    -- Query 1 (e.g., Filtering Active Users)
    SELECT User_ID, Name FROM users WHERE Status = 'Active'
), 
Second_CTE AS (
    -- Query 2 (e.g., Calculating Total Revenue)
    SELECT User_ID, SUM(Amount) as Total_Rev FROM payments GROUP BY User_ID
)

-- The Main Outer Query (Runs immediately after)
SELECT a.Name, b.Total_Rev
FROM First_CTE a
JOIN Second_CTE b 
ON a.User_ID = b.User_ID;
```

```python
query11="""
WITH MoM AS (
    SELECT emp_id, month_no, revenue, 
    LAG(revenue) OVER(PARTITION BY emp_id ORDER BY month_no ASC) AS prev_revenue
    FROM monthly_sales
)
SELECT emp_id, month_no, revenue, prev_revenue,
       (revenue - prev_revenue) AS actual_growth  -- MATH YAHAN HOGI!
FROM MoM;
"""
pd.read_sql_query(query11,conn)
```
