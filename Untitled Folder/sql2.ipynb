```python
import pandas as pd
import sqlite3

# 1. Fake Data banaya
sql_data = {
    'Emp_ID': [101, 102, 103, 104, 105, 106, 107],
    'Name': ['Ravi', 'Priya', 'Kiran', 'Amit', 'Neha', 'Rahul', 'Sonia'],
    'Department': ['IT', 'HR', 'IT', 'Sales', 'HR', 'IT', 'Sales'],
    'Salary': [55000, 62000, 48000, 75000, 50000, 90000, 45000],
    'Joining_Year': [2020, 2019, 2021, 2018, 2022, 2018, 2023]
}
emp_df = pd.DataFrame(sql_data)

# 2. SQL Engine (Connection) banaya memory mein
conn = sqlite3.connect(':memory:')

# 3. Pandas DataFrame ko SQL Table ('employees') mein convert kar diya
emp_df.to_sql('employees', conn, index=False)

print("SQL Database Ready! Table name is 'employees'.")
```

```python
import pandas as pd
import sqlite3

# 1. Customers Table (Total 4 customers)
data_cust = {
    'C_ID': [1, 2, 3, 4], 
    'Name': ['Amit', 'Neha', 'Rahul', 'Pooja'], 
    'City': ['Delhi', 'Mumbai', 'Pune', 'Delhi']
}
# 2. Orders Table (Total 5 orders)
data_ord = {
    'O_ID': [101, 102, 103, 104, 105], 
    'C_ID': [1, 2, 1, 5, 2], # Dhyan se dekh, C_ID 5 table mein nahi hai (Guest order)
    'Product': ['Laptop', 'Phone', 'Mouse', 'Tablet', 'Keyboard'], 
    'Amount': [50000, 20000, 1000, 15000, 1500]
}

df_cust = pd.DataFrame(data_cust)
df_ord = pd.DataFrame(data_ord)

# Database Setup
conn = sqlite3.connect(':memory:')
df_cust.to_sql('customers', conn, index=False)
df_ord.to_sql('orders', conn, index=False)
print("Tables 'customers' and 'orders' are ready!")
```

```python
import pandas as pd
import sqlite3

# Dummy Data Setup
sales_data = {
    'Sale_ID': ['S1', 'S2', 'S3', 'S4', 'S5', 'S6'],
    'Prod_ID': ['P1', 'P2', 'P1', 'P3', 'P2', 'P4'], 
    'Quantity': [10, 5, 20, 15, 8, 2],
    'Sale_Amount': [5000, 3000, 10000, 4500, 4800, 1000]
}

products_data = {
    'Prod_ID': ['P1', 'P2', 'P3', 'P5'], 
    'Category': ['Electronics', 'Clothing', 'Grocery', 'Books'],
    'Price': [500, 600, 300, 200]
}

df_sales = pd.DataFrame(sales_data)
df_products = pd.DataFrame(products_data)

conn = sqlite3.connect(':memory:')
df_sales.to_sql('sales', conn, index=False, if_exists='replace')
df_products.to_sql('products', conn, index=False, if_exists='replace')
print("15 LPA CTE Environment Ready!")
```

```python
WITH First_CTE AS (
    -- Query 1 (e.g., Filtering Active Users)
    SELECT User_ID, Name FROM users WHERE Status = 'Active'
), 
Second_CTE AS (
    -- Query 2 (e.g., Calculating Total Revenue)
    SELECT User_ID, SUM(Amount) as Total_Rev FROM payments GROUP BY User_ID
)

-- The Main Outer Query (Runs immediately after)
SELECT a.Name, b.Total_Rev
FROM First_CTE a
JOIN Second_CTE b 
ON a.User_ID = b.User_ID;
```

```python
query4="""
WITH Qty_Data AS(SELECT Prod_ID, SUM(Quantity) AS Total_Quantity FROM sales GROUP BY Prod_ID),
Rev_Data AS(SELECT Prod_ID,SUM(Sale_Amount) AS Total_Amount FROM sales GROUP BY Prod_ID)
SELECT q.*,r.*
FROM Qty_Data q
INNER JOIN Rev_Data r
On q.Prod_ID=r.Prod_ID
"""
pd.read_sql_query(query4,conn)
```

```python
query5="""
WITH Category_Revenue AS(SELECT sales.Prod_ID, SUM(sales.Sale_Amount) AS Total_Revenue,products.Category FROM sales
LEFT JOIN products ON sales.Prod_ID=products.Prod_ID GROUP BY products.Category)
SELECT c.*
FROM Category_Revenue c
ORDER BY Total_Revenue DESC LIMIT 1
"""
pd.read_sql_query(query5,conn)
```

```python
query6="""
WITH Sale_Tags AS(SELECT Sale_ID,Sale_Amount,
        CASE
            WHEN Sale_Amount>4000 THEN 'Big Sale'
            ELSE 'Small Sale'
        END AS Tag
    FROM sales)
SELECT *
FROM Sale_Tags
"""
pd.read_sql_query(query6,conn)
```

```python
query7="""
WITH Bulk_Products AS (SELECT Prod_Id,SUM(Quantity) AS Total_Quantity FROM sales GROUP BY Prod_ID HAVING SUM(Quantity)>10)
SELECT a.*,b.Category
FROM Bulk_Products a
INNER JOIN products b
ON a.Prod_ID=b.Prod_ID
"""
pd.read_sql_query(query7,conn)
```

```python
query8="""
WITH Product_Totals AS(SELECT Prod_ID, SUM(Sale_Amount) AS Total_Revenue FROM sales GROUP BY Prod_ID)
SELECT p.*,s.*
FROM Product_Totals p
INNER JOIN sales s
ON p.Prod_ID=s.Prod_ID
"""
pd.read_sql_query(query8,conn)
```

```python
query9="""
WITH Cat_Rev AS (SELECT sales.Prod_ID,SUM(sales.Sale_Amount) AS Total_Amount,products.Category FROM sales 
INNER JOIN products ON sales.Prod_ID=products.Prod_ID GROUP BY products.Category),
Top_Cats AS(SELECT * FROM Cat_Rev WHERE Total_Amount>5000)
SELECT *
FROM Top_Cats
"""
pd.read_sql_query(query9,conn)
```
