```python
import pandas as pd
import sqlite3

# ==========================================
# 🛠️ DATA SETUP FOR BLOCK 1
# ==========================================

# --- SQL Tables ---
products_data = {
    'product_id': ['P1', 'P2', 'P3', 'P4', 'P5', 'P101'],
    'product_name': ['Laptop', 'T-Shirt', 'Milk', 'Phone', 'Jeans', 'Premium Watch'],
    'category': ['Electronics', 'Clothing', 'Grocery', 'Electronics', 'Clothing', 'Luxury'],
    'price': [50000, 800, 50, 25000, 1200, 150000]
}
users_data = {
    'user_id': ['U1', 'U2', 'U3', 'U4', 'U5'],
    'join_date': ['2023-01-15', '2026-06-05', '2026-06-12', '2024-11-20', '2025-08-30'],
    'tier': ['Gold', 'Silver', 'Gold', 'Platinum', 'Silver'],
    'status': ['Active', 'Active', 'Inactive', 'Active', 'Active']
}
orders_data = {
    'order_id': ['O1', 'O2', 'O3', 'O4', 'O5', 'O6', 'O7'],
    'user_id': ['U1', 'U2', 'U1', 'U4', 'U5', 'U6', 'U1'], # U6 has no user record 
    'product_id': ['P1', 'P2', 'P101', 'P4', 'P2', 'P1', 'P3'],
    'amount': [50000, 1600, 150000, 25000, 2400, 50000, 50],
    'status': ['Completed', 'Shipped', 'Completed', 'Shipped', 'Pending', 'Completed', 'Shipped']
}
category_targets_data = {
    'category': ['Electronics', 'Clothing', 'Grocery', 'Luxury'],
    'target_amt': [80000, 5000, 100, 200000]
}

conn = sqlite3.connect(':memory:')
pd.DataFrame(products_data).to_sql('products', conn, index=False, if_exists='replace')
pd.DataFrame(users_data).to_sql('users', conn, index=False, if_exists='replace')
pd.DataFrame(orders_data).to_sql('orders', conn, index=False, if_exists='replace')
pd.DataFrame(category_targets_data).to_sql('category_targets', conn, index=False, if_exists='replace')

# --- Python & Pandas Data ---
prices = [120, 450, 80, 900, 310]
stock = {'P1': 50, 'P2': 12, 'P3': 0, 'P4': 85}
filenames = ['data.csv', 'image.png', 'report.pdf', 'test.csv', 'summary.docx']
df_invoice = pd.DataFrame({'quantity': [2, 5, 1], 'unit_price': [100, 50, 1000], 'shipping_fee': [20, 10, 50]})

print("✅ Block 1 Data Environment Ready!")
```

```python
q3="""
WITH Cat_Sales AS (SELECT orders.product_id,SUM(orders.amount) AS Total_Amount
,products.category FROM orders LEFT JOIN products ON orders.product_id=products.product_id GROUP BY products.category)
SELECT ca.*,cat.category,cat.target_amt
FROM Cat_Sales ca
LEFT JOIN category_targets cat
ON ca.category=cat.category
WHERE Total_Amount<target_amt
"""
pd.read_sql_query(q3,conn)
```
