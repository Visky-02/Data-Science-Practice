```python
import pandas as pd
import numpy as np

# Run this to setup Q3 data
data_q3 = {
    'Customer_ID': ['C1', 'C2', 'C3', 'C4', 'C5'],
    'Total_Spend': [15000, 4500, 25000, 8000, 12000],
    'Return_Rate': [0.05, 0.35, 0.02, 0.12, 0.25]  # 0.05 matlab 5% return rate
}

customer_df = pd.DataFrame(data_q3)
print("Raw Customer Data:\n", customer_df)
```

```python
customer_df
```

```python
customer_df['Customer_Segment']=customer_df.apply(lambda x: "VIP" if 
                                                  customer_df[(customer_df['Total_Spend']>10000) & (customer_df['Return_Rate']<0.10)] 
                                                  else "Normal")
```

```python
if customer_df['Total_Spend']>10000:
    print("VIP")
    else
        print("Standard")
    
```

```python
import numpy as np
cond1=(customer_df['Total_Spend']>10000)&(customer_df['Return_Rate']<0.10)
cond2=(customer_df['Total_Spend']>10000)&(customer_df['Return_Rate']>=0.10)
customer_df['Customer_Segment']=np.select([cond1,cond2],['VIP','High_Spend-At Risk'],default='Standard')
print(customer_df)
```

```python
# Lambda ke andar Nested If-Else lagta hai
customer_df['Segment_Lambda'] = customer_df.apply(
    lambda row: 'VIP' if (row['Total_Spend'] > 10000 and row['Return_Rate'] < 0.10) 
                else ('High Spend - At Risk' if (row['Total_Spend'] > 10000 and row['Return_Rate'] >= 0.10) 
                else 'Standard'), 
    axis=1 # IMPORTANT: Ye Python ko bolta hai "Bhai, Row by Row check kar!"
)

print(customer_df)
```
