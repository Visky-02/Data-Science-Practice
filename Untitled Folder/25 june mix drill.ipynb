```python
import pandas as pd
import sqlite3

# ==========================================
# 🛠️ DATA SETUP FOR BLOCK 1 (Q1 to Q8)
# ==========================================
conn = sqlite3.connect(':memory:')

# SQL Tables
df_users = pd.DataFrame({'user_id': [1, 2, 3, 4], 'name': ['Vishal', 'Amit', 'Neha', 'Pooja']})
df_tx = pd.DataFrame({'tx_id': [101, 102], 'user_id': [1, 3], 'amount': [5000, 1200]})
df_users.to_sql('users', conn, index=False, if_exists='replace')
df_tx.to_sql('transactions', conn, index=False, if_exists='replace')

df_employees = pd.DataFrame({'dept_id': [1, 1, 2, 2], 'salary': [60000, 80000, 50000, 40000]})
df_employees.to_sql('employees', conn, index=False, if_exists='replace')

# Pandas Data
course_revenue = {'Data Science': 150000, 'Tally Prime': 45000, 'Typing': 25000, 'Python': 80000}
attendance_pct = [85, 92, 0, 60, 78, 0, 45]
df_records = pd.DataFrame({
    'course': ['Data Science', 'Python', 'Data Science', 'Tally Prime'], 
    'amount_paid': [50000, 15000, 45000, 10000], 
    'fee_status': ['Pending', 'Pending', 'Pending', 'Pending']
})
dates_series = pd.Series(['2026-03-15', '2026-04-20', '2026-05-10'])

print("✅ Block 1 Data Environment Ready!")
```
