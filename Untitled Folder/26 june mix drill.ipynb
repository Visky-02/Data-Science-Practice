```python
import pandas as pd
import sqlite3

# ==========================================
# 🛠️ DATA SETUP FOR BLOCK 4 & 5 (Q26 to Q40)
# ==========================================
conn = sqlite3.connect(':memory:')

df_leads = pd.DataFrame({
    'lead_id': [1, 2, 3, 4, 5],
    'source': ['Google Ads', 'Meta', 'Organic', 'Google Ads', 'Meta'],
    'cost': [500, 300, 0, 600, 250],
    'converted': ['Yes', 'No', 'Yes', 'Yes', 'No']
})
df_leads.to_sql('leads', conn, index=False, if_exists='replace')

df_deals = pd.DataFrame({
    'lead_id': [1, 3, 4],
    'deal_value': [5000, 10000, 7500]
})
df_deals.to_sql('deals', conn, index=False, if_exists='replace')

scores_list = [45, 85, 30, 92]
marketing_dict = {'Meta': 5000, 'Google': 8000, 'SEO': 0}

print("✅ Final Blocks Data Ready!")
```
