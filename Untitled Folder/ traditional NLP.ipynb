```python
import pandas as pd

# Chota sa dummy dataset
data = pd.DataFrame({
    'review': ["I loved it!!! Amazing 😍", "Terrible food, very BAD.", "Good stuff."],
    'sentiment': ['Positive', 'Negative', 'Positive'] # Ye tera 'y' hai
})
```

```python
data
```

```python
import pandas as pd
import re
data['clean_review']=data['review'].str.lower()
data['clean_review']=data['clean_review'].apply(lambda x:re.sub(r'[^\w\s]',"",x))
print(data[['review','clean_review']])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer(ngram_range=(1,2))
X_matrix=tfidf.fit_transform(data['clean_review']).toarray()
print(X_matrix)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
y=data['sentiment']
X_train,X_test,y_train,y_test=train_test_split(X_matrix,y,test_size=0.2,random_state=42)
mnb=MultinomialNB()
mnb.fit(X_train,y_train)
prediction=mnb.predict(X_test)
print("accuracy of the model is : ",accuracy_score(y_test,prediction))
```
