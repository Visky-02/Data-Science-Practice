```python
titanic=pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")
```

```python
titanic.head()
```

```python
titanic.isnull().sum()
```

```python
titanic=titanic.drop('Cabin',axis=1)
```

```python
titanic.head(2)
```

```python
age_median=titanic['Age'].median()
```

```python
titanic['Age']=titanic['Age'].fillna(age_median)
```

```python
titanic.isnull().sum()
```

```python
em_mode=titanic['Embarked'].mode()[0]
```

```python
em_mode
```

```python
titanic['Embarked']=titanic['Embarked'].fillna(em_mode)
```

```python
titanic.isnull().sum()
```

```python
titanic['Family_Size']=titanic['Parch']+titanic['SibSp']
```

```python
titanic.head(2)
```

```python
titanic['Survived'].mean()*100
```

```python
titanic.groupby('Sex')['Survived'].sum()
```

```python
titanic.groupby('Sex')['Survived'].mean()*100
```

```python
titanic.groupby(['Pclass', 'Sex'])['Survived'].mean() * 100
```

```python
titanic.groupby('Survived')[['Pclass','Sex']].agg('sum')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(6,6))
sns.countplot(x='Survived',data=titanic,hue='Pclass')
plt.show()
```

```python
titanic['Age_Group']=titanic['Age'].apply(lambda x: "Child" if x<18 else "Adult")
```

```python
titanic.head(5)
```

```python
age_class1=titanic[titanic['Pclass']==1]['Age']
age_class3=titanic[titanic['Pclass']==3]['Age']
```

```python
age_class1
```

```python
from scipy import stats
t_stat,p_val=stats.ttest_ind(age_class1,age_class3)
print(f"the p value is : {p_val}")
```
