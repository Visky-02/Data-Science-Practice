```python
import pandas as pd
import sqlite3

# Dirty Tech Subscription Data
df_users = pd.DataFrame({
    'User_ID': ['U1', 'U2', 'U3', 'U4', 'U5'],
    'Plan': ['Gold', 'Silver', 'Gold', 'Bronze', 'Silver'],
    # Dhyan se dekh format: YYYY/DD/MM (Ajeeb format intentionally diya hai)
    'Join_Date_Str': ['2024/15/01', '2024/22/02', '2024/30/03', '2024/10/04', '2024/18/05'] 
})

df_payments = pd.DataFrame({
    'Pay_ID': ['P1', 'P2', 'P3', 'P4', 'P5', 'P6'],
    'User_ID': ['U1', 'U2', 'U1', 'U4', 'U2', 'U5'],
    'Amount': [5000, 2000, 5000, 500, 2000, 1500],
    'Status': ['Completed', 'Completed', 'Failed', 'Completed', 'Failed', 'Completed']
})

# SQL Database Engine
conn = sqlite3.connect(':memory:')
df_users.to_sql('users', conn, index=False, if_exists='replace')
df_payments.to_sql('payments', conn, index=False, if_exists='replace')

print("Tech Subscription Data Loaded! Ready for Batch 1.")
```

```python
query1="""
SELECT users.User_ID,users.Plan,SUM(payments.Amount) AS Collect_amount,payments.Status
FROM users
LEFT JOIN payments
ON users.User_ID=payments.User_ID
WHERE payments.Status='Completed'
GROUP BY Plan
HAVING SUM(payments.Amount)>2000
ORDER BY SUM(Payments.Amount)
"""
pd.read_sql_query(query1,conn)
```

```python
query2="""
SELECT Pay_ID,Amount
FROM payments
WHERE User_ID IN (SELECT User_ID FROM users WHERE Plan='Gold')
"""
pd.read_sql_query(query2,conn)
```

```python
query3="""
SELECT *
FROM users
WHERE User_ID NOT IN (SELECT User_ID FROM payments WHERE Status='Failed')
"""
pd.read_sql_query(query3,conn)
```
