```python
hospital=pd.get_dummies(hospital,columns=['smoker'],drop_first=True,dtype=int)
```

```python
from sklearn.preprocessing import StandardScaler
scale=StandardScaler()
hospital[['bmi','age']]=scale.fit_transform(hospital[['bmi','age']])
```
