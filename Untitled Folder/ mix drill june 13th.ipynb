```python
import pandas as pd
movies= pd.read_csv("https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/imdb_1000.csv")
```

```python
movies.columns.tolist()
```

```python
movies['genre']=movies['genre'].str.upper()
```

```python
movies
```

```python
movies['content_rating']=movies['content_rating'].replace({'R':'Restricted','PG-13':'Teens'})
```

```python
movies
```

```python
movies[(movies['star_rating']>8.5)&(movies['duration']>150)]
```

```python
movies.groupby('genre')['duration'].agg({'mean','max'})
```

```python
# movies.loc[ (Condition_1) & (Condition_2) , ['Column1', 'Column2'] ]
movies.loc[(movies['duration'] > 150) & (movies['star_rating'] > 8.5), ['title', 'genre']]
```

```python
import re
movies['actors_list']=movies['actors_list'].apply(lambda x:re.sub(r'[^\w\s]',"",x))
```

```python
movies['actors_list']=movies['actors_list'].str.lower()
```
