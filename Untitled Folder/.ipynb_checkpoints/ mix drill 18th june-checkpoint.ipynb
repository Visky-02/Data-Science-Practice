```python
import pandas as pd
import numpy as np
import sqlite3

# Dirty real-world dataset
data = {
    'Txn_ID': ['T101', 'T102', 'T103', 'T104', 'T105', 'T106', 'T107'],
    'Cust_ID': ['C1', 'C2', 'C1', 'C3', 'C2', 'C4', 'C1'],
    'Order_Date_Str': ['15-01-2024', '15-01-2024', '18-01-2024', '20-01-2024', '22-01-2024', '22-01-2024', '25-01-2024'],
    'Amount': [5000, 7000, 5000, 12000, 8000, 3000, 15000],
    'Category': ['Tech', 'Furniture', 'Tech', 'Tech', 'Furniture', 'Grocery', 'Tech'],
    'Delivery_Days': [2, 6, 1, 4, 7, 2, 5]
}

df_logistics = pd.DataFrame(data)

# SQL Setup
conn = sqlite3.connect(':memory:')
df_logistics.to_sql('logistics', conn, index=False, if_exists='replace')

print("Pandas 'df_logistics' & SQL 'logistics' table ready for 30-Q NO MERCY Drill!")
```

```python
df_logistics
```

```python
df_logistics['Order_Date_Str']=pd.to_datetime(df_logistics['Order_Date_Str'],format="%d-%m-%Y")
```

```python
df_logistics.info()
```

```python
df_logistics_clean=df_logistics[df_logistics['Category'].isin(['Tech','Furniture'])]
```

```python
df_logistics_clean.drop_duplicates(subset='Cust_ID',keep='last')
```

```python
# Dhyan se dekh nested if-else ka flow aur aakhri mein axis=1
df_logistics['Risk_Flag'] = df_logistics.apply(
    lambda x: 'High Risk' if (x['Delivery_Days'] > 5 and x['Amount'] > 6000) 
    else 'Fast Track' if x['Delivery_Days'] < 3 
    else 'Standard', 
    axis=1  # <-- YE SABSE IMPORTANT HAI ROW-WISE CHECK KE LIYE
)
print(df_logistics[['Delivery_Days', 'Amount', 'Risk_Flag']])
```

```python
# 1. Conditions ki list banao
conditions = [
    (df_logistics['Delivery_Days'] > 5) & (df_logistics['Amount'] > 6000),
    (df_logistics['Delivery_Days'] < 3)
]
# 2. Choices ki list banao
choices = ['High_Risk', 'Fast_Track']

# 3. Apply kar do
df_logistics['Risk_Flag'] = np.select(conditions, choices, default='Standard')
print(df_logistics[['Delivery_Days', 'Amount', 'Risk_Flag']])
```

```python
# Har 'x' (category) ke liye ye check hoga, bracket gaur se dekh!
df_logistics['Cat_Code'] = ['T' if x == 'Tech' else 'F' if x == 'Furniture' else 'O' for x in df_logistics['Category']]
print(df_logistics[['Category', 'Cat_Code']])
```

```python
# Poore DF ki jagah sirf 'Category' Series par apply kiya
# Ab 'x' ka matlab poori row nahi, sirf Category ka naam (jaise 'Tech') hai
df_logistics['Cat_Code'] = df_logistics['Category'].apply(
    lambda x: 'T' if x == 'Tech' else 'F' if x == 'Furniture' else 'O'
)
print(df_logistics[['Category', 'Cat_Code']])
```
