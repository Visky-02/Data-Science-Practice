```python
import pandas as pd
import sqlite3

# ==========================================
# 🛠️ DATA SETUP FOR BLOCK 1 (Q1 to Q10)
# ==========================================
conn = sqlite3.connect(':memory:')

df_customers = pd.DataFrame({
    'cust_id': [101, 102, 103, 104, 105],
    'email': ['vishal@tech.com', 'amit@sales.in', 'neha@hr.net', 'pooja@tech.com', 'karan@sales.in'],
    'status': ['Active', 'Inactive', 'Active', 'Active', 'Inactive'],
    'spend': [5000, 0, 12000, 8000, 0]
})
df_customers.to_sql('customers', conn, index=False, if_exists='replace')

df_orders = pd.DataFrame({
    'order_id': [1, 2, 3],
    'cust_id': [101, 103, 104],
    'amount': [5000, 12000, 8000]
})
df_orders.to_sql('orders', conn, index=False, if_exists='replace')

raw_string = "Order-9921-Delivered"
spend_list = [5000, 0, 12000, 8000, 0]

print("✅ Rust-Breaker Block 1 Data Ready!")
```

```python
query1="""
SELECT cust_id,email
FROM customers
WHERE cust_id NOT IN (SELECT cust_id FROM orders)
"""
pd.read_sql_query(query1,conn)
```

```python
query2="""
SELECT SUM(CASE WHEN status='Active' THEN 1 END) AS total_active, SUM(CASE WHEN status='Churned' THEN 1 END) AS total_inactive
FROM customers
"""
pd.read_sql_query(query2,conn)
```

```python
query3="""
WITH rank AS(SELECT cust_id,status,spend,
DENSE_RANK() OVER(PARTITION BY status ORDER BY spend DESC) AS spend_rank FROM customers)
SELECT *
FROM rank
"""
pd.read_sql_query(query3,conn)
```

```python
import pandas as pd
import sqlite3

# ==========================================
# 🛠️ DATA SETUP FOR BLOCK 2 (Q11 to Q20)
# ==========================================
conn = sqlite3.connect(':memory:')

df_products = pd.DataFrame({
    'prod_id': [201, 202, 203, 204],
    'category': ['Electronics', 'Furniture', 'Electronics', 'Furniture'],
    'price': [15000, 8000, 22000, 4500]
})
df_products.to_sql('products', conn, index=False, if_exists='replace')

df_sales2 = pd.DataFrame({
    'sale_id': [1, 2, 3, 4, 5],
    'prod_id': [201, 202, 201, 205, 203],
    'qty': [2, 1, 1, 5, 2]
})
df_sales2.to_sql('sales2', conn, index=False, if_exists='replace')

dates_list = ['2026-06-01', '2026-06-15', '2026-06-28']
budget = {'IT': 500, 'HR': 200, 'Sales': 700}

print("✅ Block 2 Data Ready!")
```

```python
query4="""
SELECT s.sale_id,p.prod_id
FROM sales2 s
LEFT JOIN products p
ON s.prod_id=p.prod_id
WHERE p.prod_id IS NULL 
"""
pd.read_sql_query(query4,conn)
```

```python
query5="""
SELECT prod_id, SUM(qty) AS total_qty
FROM sales2
GROUP BY prod_id
HAVING SUM(qty)>2
"""
pd.read_sql_query(query5,conn)
```

```python
query6="""
SELECT p.category,SUM(p.price*s.qty) AS total_revenue
FROM products p
LEFT JOIN sales2 s
ON p.prod_id=s.prod_id
GROUP BY p.category
"""
pd.read_sql_query(query6,conn)
```

```python
query7="""
WITH a AS (SELECT category,price , ROW_NUMBER() OVER(PARTITION BY category ORDER BY price DESC) AS top_cat FROM products)
SELECT *
FROM a
WHERE top_cat=1
"""
pd.read_sql_query(query7,conn)
```

```python
import pandas as pd
import sqlite3

# ==========================================
# 🛠️ DATA SETUP FOR BLOCK 3 (Q21 to Q30)
# ==========================================
conn = sqlite3.connect(':memory:')

df_inventory2 = pd.DataFrame({
    'inv_id': [1, 2, 3, 4, 5],
    'item_name': ['Desk', 'Chair', 'Lamp', 'Desk', 'Monitor'],
    'stock': [10, pd.NA, 5, pd.NA, 20] # pd.NA means missing value
})
df_inventory2.to_sql('inventory2', conn, index=False, if_exists='replace')

df_orders2 = pd.DataFrame({
    'order_id': [101, 102, 103],
    'inv_id': [1, 3, 5],
    'qty_sold': [2, 1, 4]
})
df_orders2.to_sql('orders2', conn, index=False, if_exists='replace')

codes_list = ['A-100', 'B-200', 'C-300']
salary_dict = {'Dev': 80000, 'QA': 45000, 'Ops': 60000}

print("✅ Block 3 Data Ready!")
```

```python
df_inventory2
```

```python
df_inventory2[df_inventory2['item_name'].isin(['Chair','Lamp'])]
```

```python
[x.split('-')[1] for x in codes_list]
```

```python
query8="""
SELECT i.inv_id,i.item_name,o.qty_sold,COALESCE(o.qty_sold,0) AS clean_qty_sold
FROM inventory2 i
LEFT JOIN orders2 o
ON i.inv_id=o.inv_id
"""
pd.read_sql_query(query8,conn)
```

```python
df_inventory2['stock']=df_inventory2['stock'].fillna(0,inplace=True)
```

```python
df_inventory2
```

```python
query9="""
SELECT item_name, COUNT(item_name) AS no_of_item
FROM inventory2
GROUP BY item_name
HAVING COUNT(item_name)>1
"""
pd.read_sql_query(query9,conn)
```

```python
df_inventory2.loc[df_inventory2['item_name']=='Desk', 'stock']=df_inventory2['stock']-2
```

```python
df_inventory2
```

```python
query10="""
SELECT i.item_name,SUM(i.stock-o.qty_sold) AS Remaining
FROM inventory2 i
LEFT JOIN orders2 o
ON i.inv_id=o.inv_id
GROUP BY i.item_name
"""
pd.read_sql_query(query10,conn)
```

```python
query11="""
WITH b AS(SELECT item_name, stock, DENSE_RANK() OVER(ORDER BY stock DESC) stock_rank FROM inventory2)
SELECT *
FROM b
WHERE stock_rank<=2
"""
pd.read_sql_query(query11,conn)
```
