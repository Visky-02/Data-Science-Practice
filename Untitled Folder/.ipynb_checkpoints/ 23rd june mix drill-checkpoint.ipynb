```python
import pandas as pd
import sqlite3
import numpy as np

# 1. Students Data
students_data = {
    'Stud_ID': ['S1', 'S2', 'S3', 'S4', 'S5', 'S6'],
    'Course': ['Data Analytics', 'Data Science', 'Data Analytics', 'Power BI', 'Data Science', 'Power BI'],
    'Join_Date': ['15-06-2026', '20-05-2026', '15-06-2026', '01-04-2026', '10-06-2026', '22-05-2026'],
    'Fee_Paid': [50000, 75000, 50000, 15000, 75000, 15000]
}

# 2. Course Views Data
views_data = {
    'View_ID': ['V1', 'V2', 'V3', 'V4', 'V5', 'V6', 'V7'],
    'Stud_ID': ['S1', 'S2', 'S1', 'S4', 'S5', 'S7', 'S2'], # Dhyan de S7 ek unregistered viewer hai
    'View_Mins': [45, 120, 30, 15, 90, 10, 60],
    'Device': ['Mobile', 'Laptop', 'Laptop', 'Mobile', 'Laptop', 'Mobile', 'Tablet']
}

df_students = pd.DataFrame(students_data)
df_views = pd.DataFrame(views_data)

# SQL Database Engine Connection
conn = sqlite3.connect(':memory:')
df_students.to_sql('students', conn, index=False, if_exists='replace')
df_views.to_sql('views', conn, index=False, if_exists='replace')

print("EdTech Database Loaded! Ready for 30-Q Drill.")
```

```python
query1="""
SELECT Stud_ID, Course
FROM students
WHERE Stud_ID NOT IN (SELECT Stud_ID FROM students)
"""
pd.read_sql_query(query1,conn)
```

```python
df_students['Join_Date']=pd.to_datetime(df_students['Join_Date'],format='%d-%m-%Y')
```

```python
df_students['Months']=df_students['Join_Date'].dt.month
```

```python
query2="""
SELECT students.Stud_ID,students.Course,views.Device,SUM(views.View_Mins) AS Total_View_Min
FROM students
LEFT JOIN views
ON students.Stud_ID=views.Stud_ID
WHERE views.Device=='Laptop'
Group BY Course
HAVING SUM(views.View_Mins)>100
"""
pd.read_sql_query(query2,conn)
```

```python
df_students['Fee_Tag']=['Premium' if x>60000 else 'Standard' if x>20000 else 'Basic' for x in df_students['Fee_Paid']]
```

```python
query3="""
SELECT Stud_ID, Course, Fee_Paid
FROM students
WHERE Fee_Paid>(SELECT AVG(Fee_Paid) AS Avg_fee FROM students)
"""
pd.read_sql_query(query3,conn)
```
