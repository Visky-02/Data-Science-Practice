```python
import pandas as pd
import sqlite3

# Naye Variables banate hain
emp_data = {
    'Emp_ID': [1, 2, 3, 4, 5, 6],
    'Name': ['Rahul', 'Priya', 'Amit', 'Neha', 'Vikram', 'Sonia'],
    'Salary': [75000, 60000, 85000, 50000, 72000, 90000],
    'Dept_ID': [101, 102, 101, 103, 102, 999] 
}

dept_data = {
    'Dept_ID': [101, 102, 103, 104],
    'Dept_Name': ['IT', 'Sales', 'HR', 'Marketing'],
    'Location': ['Bangalore', 'Mumbai', 'Delhi', 'Pune']
}

# Pandas DataFrames banana
Employees_df = pd.DataFrame(emp_data)
Departments_df = pd.DataFrame(dept_data)

# SQL Database mein replace karna (Taki purana kachra saaf ho jaye)
conn = sqlite3.connect(':memory:')
Employees_df.to_sql('Employees', conn, index=False, if_exists='replace')
Departments_df.to_sql('Departments', conn, index=False, if_exists='replace')

print("Old tables destroyed. New Tables Ready for JOINS, CEO!")
```
