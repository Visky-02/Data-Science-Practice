```python
import pandas as pd
movies=pd.read_csv('https://raw.githubusercontent.com/YBI-Foundation/Dataset/main/Movies%20Recommendation.csv')
```

```python
features=['Movie_Genre','Movie_Keywords','Movie_Tagline','Movie_Cast','Movie_Director']
for x in features:
    movies[x]=movies[x].fillna('')
movies['Combined_Features']=movies['Movie_Genre']+ ' ' + movies['Movie_Keywords'] + ' ' + movies['Movie_Tagline'] + ' ' + movies['Movie_Cast'] + ' ' + movies['Movie_Director']
```

```python
movies
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer=TfidfVectorizer(stop_words='english',max_features=5000)
feature_vector=vectorizer.fit_transform(movies['Combined_Features'])
```

```python
print(feature_vector)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity=cosine_similarity(feature_vector)
print(similarity)
print('matrix shape: ',similarity.shape)
```

```python
movies.columns
```

```python
import difflib
list_of_all_titles=movies['Movie_Title'].tolist()
print(list_of_all_titles[:5])
```

```python
import difflib
movie_name='Iron Man'
find_closet_match=difflib.get_close_matches(movie_name,list_of_all_titles)
print("closest match found: ",find_closet_match)
close_match=find_closet_match[0]
print("final selected movie is: ",close_match)
```

```python
import pickle
pickle.dump(movies.to_dict(),open('movie_dict.pkl','wb'))
pickle.dump(similarity,open('similarity.pkl','wb'))
print('Model and Data both are successfully packed')
```
