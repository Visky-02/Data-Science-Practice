```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
import warnings
warnings.filterwarnings('ignore')
df = pd.read_excel('Online retail.xlsx', header=None)
```

```python
transactions = []
for index, row in df.iterrows():
    items = str(row[0]).split(',')
    cleaned_items = [item.strip().replace('"', '') for item in items if item.strip()]
    if cleaned_items:
        transactions.append(cleaned_items)

print(f"Total number of transactions: {len(transactions)}")
```

```python
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
encoded_df = pd.DataFrame(te_ary, columns=te.columns_)

print("Data Shape after encoding:", encoded_df.shape)
encoded_df.head()
```

```python
from mlxtend.frequent_patterns import apriori, association_rules
frequent_itemsets = apriori(encoded_df, min_support=0.01, use_colnames=True)
print("Number of frequent itemsets found:", len(frequent_itemsets))
```

```python
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.2)
strong_rules = rules[rules['lift'] > 1.0].sort_values(by='lift', ascending=False)
print(f"Total strong rules extracted: {len(strong_rules)}")
```

```python
strong_rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(10)
```

### Analysis and Interpretation (Business Insights)
Based on the generated rules, we can observe distinct customer purchasing behaviors:
* **Strong Product Correlations:** Customers buying items like 'herb & pepper' or 'ground beef' are highly likely to purchase 'mineral water' and 'spaghetti'.
* **Bundling Opportunities:** Products that frequently appear together with a high Lift (>1) indicate a strong positive relationship. The retail store can place these products adjacent to each other on shelves or create combo offers (e.g., pasta & shrimp combos).

### Interview Questions

**1. What is lift and why is it important in Association rules?**
* **Lift** measures the strength of association between two items. It is the ratio of the observed support to that expected if the two items were independent. 
* **Importance:** A Lift > 1 means the items are positively correlated (buying one increases the chances of buying the other). A Lift of 1 means no association, and < 1 means they are substitutes (negatively correlated). It is crucial because Confidence alone can be misleading if the consequent item is just generally very popular.

**2. What is Support and Confidence? How do you calculate them?**
* **Support:** It measures how frequently an item or itemset appears in the dataset. 
  * *Formula:* `Support(A) = (Number of transactions containing A) / (Total number of transactions)`
* **Confidence:** It measures the likelihood of item B being purchased when item A is purchased. 
  * *Formula:* `Confidence(A → B) = Support(A ∪ B) / Support(A)`

**3. What are some limitations or challenges of Association rules mining?**
* **Computational Cost:** For large datasets with many unique items, generating all possible itemsets requires massive memory and processing power.
* **Spurious Rules:** Setting the minimum thresholds too low can generate thousands of meaningless or obvious rules (e.g., "Bread → Milk").
* **Rare Items:** Setting the support threshold too high might cause the algorithm to miss interesting but rare item associations.
