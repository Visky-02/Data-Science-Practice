```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
df = pd.read_csv('Alphabets_data.csv')
```

```python
df.isnull().sum()
```

```python
df['letter'].unique()
```

```python
le = LabelEncoder()
df['letter'] = le.fit_transform(df['letter'])
X = df.drop('letter', axis=1)
y = df['letter']
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report
ann = MLPClassifier(hidden_layer_sizes=(50, 50), max_iter=500, random_state=42)
ann.fit(X_train, y_train)
y_pred = ann.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}%")
```

```python
from sklearn.model_selection import GridSearchCV
param_grid = {
    'hidden_layer_sizes': [(50, 50), (100,)],
    'activation': ['relu', 'tanh'],
    'learning_rate_init': [0.01, 0.001]
}
grid = GridSearchCV(MLPClassifier(max_iter=500, random_state=42), param_grid, cv=3, n_jobs=-1)

print("Tuning chal rahi hai, thoda time lagega...")
grid.fit(X_train, y_train)
print("Best Parameters:", grid.best_params_)
best_model = grid.best_estimator_
y_pred_tuned = best_model.predict(X_test)
print(f"Tuned Accuracy: {accuracy_score(y_test, y_pred_tuned) * 100:.2f}%")
```

```python
from sklearn.metrics import classification_report

print("--- Basic Model Report ---")
print(classification_report(y_test, y_pred))

print("\n--- Tuned Model Report ---")
print(classification_report(y_test, y_pred_tuned))
```
