## Project Summary & Actionable Recommendations

**Business Objective:**
Identify the root causes behind the 26.5% customer churn rate and propose data-driven solutions to improve customer retention.

**1. Key Findings (The Problems):**
* **The Contract Trap:** Month-to-month users are highly volatile, churning at ~43%. Customers on 1-year or 2-year contracts have almost zero churn.
* **The Premium Product Failure:** Fiber Optic is our premium service, but users who buy it *without* Tech Support have a massive 49% churn rate. 
* **The Danger Zone (Time):** The highest churn density is in the first 12 months (0-1 Year bucket). Customers leave very early in their lifecycle.
* **Price Sensitivity:** The churn spikes aggressively when the Monthly Charges hit the $70 - $110 range. Expensive plans are driving customers away.

**2. Actionable Recommendations (The Solutions):**
* **Incentivize Long-Term Contracts:** Offer a one-time discount or a free OTT subscription to Month-to-month customers if they upgrade to a 1-year contract.
* **Bundle Tech Support:** Make basic Tech Support completely FREE or mandatory for the first 6 months for all Fiber Optic customers to reduce frustration.
* **Fix the Onboarding Experience:** Create a proactive '90-Day Retention Program' for new customers, including feedback calls and basic issue-resolution check-ins.
* **Loyalty Perks for Premium Users:** Customers paying $70+ should be enrolled in a premium loyalty program to increase their perceived value and ROI.
