```python
import pandas as pd
ecom=pd.read_csv("online_retail.csv")
```

```python
ecom
```

```python
 ecom['Total_revenue']=ecom['Quantity']*ecom['Price']
```

```python
ecom.isnull().sum()
```

```python
ecom.rename(columns={'Customer ID': 'Customer_ID'}, inplace=True)
```

```python
ecom.dropna(inplace=True)
```

```python
ecom.isnull().sum()
```

```python
ecom.info()
```

```python
ecom['InvoiceDate'] = pd.to_datetime(ecom['InvoiceDate'])
```

```python
ecom.info()
```

```python
latest_date = ecom['InvoiceDate'].max()
print(latest_date)
```

```python
import datetime as dt
snapshot_date = latest_date + dt.timedelta(days=1)
print(snapshot_date)
```

```python
rfm=ecom.groupby('Customer_ID').agg({'InvoiceDate': 'max','Invoice': 'nunique','Total_revenue': 'sum'})
```

```python
rfm.rename(columns={'InvoiceDate': 'Recency_Date','Invoice': 'Frequency','Total_revenue': 'Monetary'},inplace=True)
```

```python
rfm.head(2)
```

```python
rfm.info()
```

```python
snapshot_date
```

```python
rfm['Recency'] = (snapshot_date - rfm['Recency_Date']).dt.days
```

```python
rfm
```

```python
rfm.drop(columns=['Recency_Date'], inplace=True)
```

```python
rfm = rfm[['Recency', 'Frequency', 'Monetary']]
```

```python
rfm.head(2)
```

```python
rfm['R_Score'] = pd.qcut(rfm['Recency'], q=4, labels=[4, 3, 2, 1])
```

```python
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'), q=4, labels=[1, 2, 3, 4])
```

```python
rfm['M_Score'] = pd.qcut(rfm['Monetary'], q=4, labels=[1, 2, 3, 4])
```

```python
rfm['RFM_Score'] = rfm['R_Score'].astype(str) + rfm['F_Score'].astype(str) + rfm['M_Score'].astype(str)
```

```python
rfm
```

```python
def segment_customer(score_str):
    r = int(score_str[0])
    f = int(score_str[1])
    m = int(score_str[2])
    
    if r >= 3 and f >= 3 and m >= 3:
        return 'VIP / Champions'
    elif r >= 3 and f <= 2:
        return 'New & Potential'
    elif r <= 2 and f >= 3 and m >= 3:
        return 'At Risk (Loyal but leaving)'
    elif r <= 2 and f <= 2:
        return 'Lost Customers'
    else:
        return 'Regulars'
```

```python
rfm['Customer_Segment'] = rfm['RFM_Score'].apply(segment_customer)
```

```python
rfm['Customer_Segment'].value_counts()
```

## Executive Summary:

Analyzed a raw UK-based E-commerce dataset of 1 Million+ transaction records. Cleaned data anomalies (missing Customer IDs), engineered a 'Gross Revenue' feature, and developed a robust RFM (Recency, Frequency, Monetary) model using Pandas Quantile Cuts to segment ~6,000 unique customers.

## Conclusion (The Data Story):

The business is heavily polarized. While we have a strong core of 1,155 VIP/Champion customers driving consistent revenue, there is a massive churn issue with 1,782 Lost Customers. More critically, 756 customers are 'At Risk'—these are people who used to buy frequently and spend heavily, but haven't visited the store recently.

## The Business Solution (Actionable Strategies):

For VIPs / Champions: Introduce a premium "Loyalty Tier" with early access to new products. No heavy discounts needed; they value exclusivity.

For At-Risk (Loyal but leaving): This is where the marketing budget should go. Send highly personalized "We Miss You" emails with a strong, time-sensitive discount (e.g., 20% off) to reactivate them before they are lost forever.

For New & Potential: Trigger automated onboarding emails and cross-sell/upsell product recommendations based on their first purchase.

For Lost Customers: Stop spending marketing ad dollars on them. Move them to a low-cost, generic newsletter list.
